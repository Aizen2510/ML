```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
# dataset 
dataset = np.array([
    [1,1,0], [2,10,0], [3,5,0],[4,2,0],
    [5,8,0], [6,4,0], [7,7,0], [8,3,0],
    [9,9,1], [10,6,0], [11,11,0], [12,12,0],
    [13,13,1], [14,14,0], [15,15,0], [16,16,0],
    [17,17,0], [18,18,0], [19,19,0], [20,20,0],
    [21,21,0], [22,22,1], [23,23,0], [24,24,0],
])
```

```python
# chia 2 nhom 
gr_a = dataset[dataset[:,2] == 0]
gr_b = dataset[dataset[:,2] == 1]
```

```python
# bieu do truc quan 
plt.scatter(gr_a[:,0], gr_a[:,1], color='blue', label='Group A')
plt.scatter(gr_b[:,0], gr_b[:,1], color='red', label='Group B')

plt.title('Decision Tree Dataset Visualization')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid()
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree


# Chia du lieu thanh 2 phan
X = dataset[:, :2]  # Chon 2 cot dau tien la dac trung
y = dataset[:, 2]   # Cot cuoi cung la nhan

# Tao mo hinh cay quyet dinh
clf = DecisionTreeClassifier(random_state=0)
clf.fit(X, y)
```

```python
plt.figure(figsize=(12, 8))
plot_tree(clf, filled=True, feature_names=['Feature 1', 'Feature 2'], class_names=['Group A', 'Group B'])
plt.title('Decision Tree Visualization')
plt.show()      
```
