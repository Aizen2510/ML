```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
dataset = np.array([
    [-10,0],
    [-9,0],
    [-8,0],
    [-7,0],
    [-6,0],
    [-5,0],
    [5,1],
    [6,1],
    [7,1],
    [8,1],
    [9,1],
    [10,1]
])
```

```python
negetive_class = dataset[dataset[:,1] == 0]
positive_class = dataset[dataset[:,1] == 1]

plt.scatter(negetive_class[:,0], negetive_class[:,1], color='blue', label='Negative Class')
plt.scatter(positive_class[:,0], positive_class[:,1], color='red', label='Positive Class')
plt.title('Logistic Regression Dataset Visualization')
plt.xlabel('Feature')
plt.ylabel('Label')
plt.legend()
plt.grid()
plt.show()
```

```python
def get_prediction(m,b,x):
    return 1 / (1 + np.exp(-(m*x + b)))
def get_loss(y_true, y_pred):
    return -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))
def get_gradient(m,b,x,y, y_hat):
    m_gradient = np.mean((y_hat - y) * x)
    b_gradient = np.mean(y_hat - y)
    return m_gradient, b_gradient

def get_accurecy(y,y_hat):
    return ((y_hat >= 0.5).astype(int)== y).sum()/y.shape[0]
```
